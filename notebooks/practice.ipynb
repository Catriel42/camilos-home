```python
def score_clasification(score: int) -> str : 
    if score >= 90:
        return "Excelent"
    elif score >= 75:
        return "penetubre"
    elif score >= 51:
        return "hmmm ya, no me hagas perder el tiempo"
    elif score < 51:
        return "Reprobado"
```

```python
score_clasification(74)
```

```python
df["califications"] = df["math score"].apply(score_clasification)
```
